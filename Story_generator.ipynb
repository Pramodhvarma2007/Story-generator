```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
```

```python
for m in client.models.list():
  print(m.name)
```

```python
MODEL = "gemini-3.1-flash-lite"
```

create story function**

```python
def generate_story(topic, genre, length_words=300):

    system_prompt = """
    You are a creative short story writer.
    Write an interesting and simple story based on the
    topic and genre given by the user.
    """

    user_prompt = f"""
    Write a short story.

    Topic: {topic}
    Genre: {genre}
    Length: approximately {length_words} words
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=user_prompt,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=1,
            max_output_tokens=500,
            thinking_config=types.ThinkingConfig(
                thinking_level="low"
            )
        )
    )

    return response.text
```

taking input

```python
topic = input("Enter story topic: ")
genre = input("Enter story genre: ")
length = int(input("Enter approximate story length: "))

story = generate_story(topic, genre, length)

print("\n===== GENERATED STORY =====\n")
print(story)
```
